```python
import pandas as pd
df=pd.read_csv(r"C:\Users\venka\Downloads\job_title_des.csv.zip")
print(df.head())
```

```python
print("Columns in the dataset:")
print(df.columns.tolist())
print("------"*2)

print("data info")
df.info()
print("------"*2)

print("Missing values:")
print(df.isnull().sum())
print("------"*2)

print("Number of duplicate records:", df.duplicated().sum())
```

```python
df = df.drop(columns=["Unnamed: 0"])

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns)

df.head()
```

```python
print("Number of unique job titles:", df["Job Title"].nunique())

print("\nJob title distribution:")
print(df["Job Title"].value_counts())
```

```python
for i in range(3):
    print("Job Title:", df["Job Title"].iloc[i])
    print("Job Description:")
    print(df["Job Description"].iloc[i])
    print("-" * 80)
```

```python
df["Cleaned_Text"] = df["Job Description"].str.lower()

df[["Job Description", "Cleaned_Text"]].head()
```

```python
df["Cleaned_Text"] = df["Cleaned_Text"].str.replace(
    r"[^a-zA-Z\s]", "", regex=True
)

df["Cleaned_Text"] = df["Cleaned_Text"].str.replace(
    r"\s+", " ", regex=True
).str.strip()

df[["Job Description", "Cleaned_Text"]].head()
```

```python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")

from nltk.tokenize import word_tokenize

df["Tokens"] = df["Cleaned_Text"].apply(word_tokenize)

df[["Job Title", "Tokens"]].head()
```

```python
nltk.download("stopwords")

from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))



df["Tokens"] = df["Tokens"].apply(
    lambda words: [word for word in words if word not in stop_words]
)



df[["Job Title", "Tokens"]].head()
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

df["Tokens"] = df["Tokens"].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)

df[["Job Title", "Tokens"]].head()
```

```python
df["Processed_Text"] = df["Tokens"].apply(lambda words: " ".join(words))

df[["Job Title", "Processed_Text"]].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer


bigram_vectorizer = CountVectorizer(ngram_range=(2, 2))


bigram_matrix = bigram_vectorizer.fit_transform(df["Processed_Text"])


bigram_words = bigram_vectorizer.get_feature_names_out()

bigram_counts = bigram_matrix.sum(axis=0).A1


bigram_freq = pd.DataFrame({
    "Bigram": bigram_words,
    "Frequency": bigram_counts
})


bigram_freq = bigram_freq.sort_values(
    by="Frequency",
    ascending=False
)


bigram_freq.head(20)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer



tfidf = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=5000
)


X = tfidf.fit_transform(df["Processed_Text"])

print("Feature matrix shape:", X.shape)
```

```python
from sklearn.preprocessing import LabelEncoder


label_encoder = LabelEncoder()


y = label_encoder.fit_transform(df["Job Title"])


print("Encoded job categories:")
print(y[:10])

print("\nJob categories:")
print(label_encoder.classes_)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training labels shape:", y_train.shape)
print("Testing labels shape:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression


model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")


y_pred = model.predict(X_test)

print("Predictions completed successfully!")
print("First 10 predicted labels:")
print(y_pred[:10])
```

```python
predicted_job_titles = label_encoder.inverse_transform(y_pred)

print("First 10 predicted job titles:")
print(predicted_job_titles[:10])
```

```python
from sklearn.metrics import accuracy_score


accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Model Accuracy (%):", accuracy * 100)
```

```python
from sklearn.metrics import classification_report

report = classification_report(
    y_test,
    y_pred,
    target_names=label_encoder.classes_
)

print(report)
```

```python
new_job_description = """
Looking for a developer with strong Python experience,
Django framework knowledge, REST API development,
SQL and JSON skills. Experience in backend development
and web application development is preferred.
"""

# Apply the same preprocessing steps

new_text = new_job_description.lower()

new_text = pd.Series([new_text]).str.replace(
    r"[^a-zA-Z\s]", "", regex=True
).iloc[0]

new_text = pd.Series([new_text]).str.replace(
    r"\s+", " ", regex=True
).str.strip().iloc[0]

# Tokenization

new_tokens = word_tokenize(new_text)

# Stop-word removal

new_tokens = [
    word for word in new_tokens
    if word not in stop_words
]

# Lemmatization

new_tokens = [
    lemmatizer.lemmatize(word)
    for word in new_tokens
]

# Convert tokens back to text

new_processed_text = " ".join(new_tokens)

# Convert new text into TF-IDF features
new_tfidf = tfidf.transform([new_processed_text])

# Predict job category

new_prediction = model.predict(new_tfidf)

# Convert numerical prediction to job title

predicted_job = label_encoder.inverse_transform(new_prediction)

print("New Job Description:")
print(new_job_description)

print("\nProcessed Text:")
print(new_processed_text)

print("\nPredicted Job Title:")
print(predicted_job[0])
```
